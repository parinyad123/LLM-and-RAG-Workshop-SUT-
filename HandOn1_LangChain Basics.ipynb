## 1.5 MessagesPlaceholder - For Conversation History

### 1.5.1: The Stateless Challenge (เมื่อ AI ไม่มีความจำ)

```python
# 1. Define Stateless Chat Template
stateless_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful barista assistant."),
    ("user", "{question}")
])

# --- Turn 1: Initial Order ---
question1 = "I'd like to order a Hot Americano."
# Format template and invoke model
response1 = llm.invoke(stateless_template.invoke({"question": question1}))

print(f"User: {question1}")
display(Markdown(response1.content))

# --- Turn 2: Follow-up (Missing Context) ---
question2 = "Can you make it less sweet?"
# Prompt contains only Question 2; model has no memory of the Americano
response2 = llm.invoke(stateless_template.invoke({"question": question2}))

print(f"User: {question2}")
display(Markdown(response2.content))

# --- Turn 3: Request for Total ---
question3 = "How much is it?"
# Prompt contains only Question 3; model doesn't know what is being priced
response3 = llm.invoke(stateless_template.invoke({"question": question3}))

print(f"User: {question3}")
display(Markdown(response3.content))
```

```python
# 1. Define Template with MessagesPlaceholder for history
conversation_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful barista assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("user", "{question}")
])

# Initialize empty list to store conversation context
chat_history = []

# --- Turn 1: Initial Order ---
question1 = "I'd like to order a Hot Americano."
# Pass current history and new question
response1 = llm.invoke(conversation_template.invoke({"chat_history": chat_history, "question": question1}))

print(f"User: {question1}")
display(Markdown(response1.content))

# Manually append Turn 1 to history
chat_history.append(HumanMessage(content=question1))
chat_history.append(AIMessage(content=response1.content))

# --- Turn 2: Follow-up (Contextual) ---
question2 = "Can you make it less sweet?"
# Model now sees the Americano order from chat_history
response2 = llm.invoke(conversation_template.invoke({"chat_history": chat_history, "question": question2}))

print(f"User: {question2}")
display(Markdown(response2.content))

# Append Turn 2 to history
chat_history.append(HumanMessage(content=question2))
chat_history.append(AIMessage(content=response2.content))

# --- Turn 3: Final Confirmation ---
question3 = "How much is it?"
# Model understands 'it' refers to the Hot Americano
response3 = llm.invoke(conversation_template.invoke({"chat_history": chat_history, "question": question3}))

print(f"User: {question3}")
display(Markdown(response3.content))
```

# Part 2: LangChain Expression Language (LCEL) Chains

## 2.1 What is LCEL?

```python
# --- ❌ Traditional Approach (Manual Steps) ---
# Define the prompt template
template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "{question}")
])

# Step 1: Manually format the template into a prompt object
prompt = template.invoke({"question": "What is 2+2?"})

# Step 2: Manually pass that prompt object to the LLM
response = llm.invoke(prompt)

# Step 3: Manually extract the text content from the result object
answer = response.content

print("❌ Traditional Approach:")
print(f"Answer: {answer}\n")


# --- ✅ The LCEL Approach (Declarative Pipeline) ---
# Define a "Chain" where data flows from Template -> LLM -> Parser
# The '|' operator pipes the output of one step into the input of the next
chain = template | llm | StrOutputParser()

# Execute the entire workflow in a single call
# No need to manage intermediate variables like 'prompt' or 'response'
answer = chain.invoke({"question": "What is 2+2?"})

print("✅ With LCEL Approach:")
print(f"Answer: {answer}")
```

## 2.2 Basic Chain Pattern

```python

 # --- ท่อที่ 1: นักวิเคราะห์ตลาด ---
analysis_prompt = ChatPromptTemplate.from_messages([
    ("system", "คุณเป็นนักวิเคราะห์การตลาด หน้าที่ของคุณคือระบุกลุ่มเป้าหมายหลัก 1 กลุ่มสำหรับสินค้าที่ได้รับ"),
    ("user", "สินค้าคือ: {product_name}")
])
analysis_chain = analysis_prompt | llm | StrOutputParser()

# --- ท่อที่ 2: นักเขียนโฆษณาเจาะกลุ่ม ---
copy_prompt = ChatPromptTemplate.from_messages([
    ("system", "คุณเป็นนักเขียนโฆษณา หน้าที่ของคุณคือเขียนโพสต์ Facebook สำหรับกลุ่มเป้าหมายที่ระบุ"),
    ("user", "กลุ่มเป้าหมาย: {audience}, สินค้าคือ: {product_name}")
])
copy_chain = copy_prompt | llm | StrOutputParser()

# --- การรวมท่อ (Sequential Workflow) ---
# 1. วิเคราะห์กลุ่มเป้าหมายก่อน
product = "รองเท้าวิ่งสำหรับผู้สูงอายุ"
audience = analysis_chain.invoke({"product_name": product})

# 2. นำกลุ่มเป้าหมายที่ได้ มาเขียนโฆษณาต่อ
final_post = copy_chain.invoke({"audience": audience, "product_name": product})

print(f"วิเคราะห์กลุ่มเป้าหมาย: {audience}")
print(f"โพสต์ที่ได้: {final_post}")
```

## 2.3 Sequential Chains - Multi-Step Workflows

```python
"""
Sequential Chain: Output of step N → Input of step N+1
Example: Generate summary → Extract key points
"""

# --- STEP 1: Summarization Pipeline ---
# Define template for condensing input text
summarize_template = ChatPromptTemplate.from_messages([
    ("system", "Summarize the following text in 2-3 sentences."),
    ("user", "{text}")
])

# --- STEP 2: Extraction Pipeline ---
# Define template to extract highlights from a summary
extract_template = ChatPromptTemplate.from_messages([
    ("system", "From the summary, extract 3 key bullet points."),
    ("user", "{summary}")
])

# Build modular component chains using LCEL
summarize_chain = summarize_template | llm | StrOutputParser()
extract_chain = extract_template | llm | StrOutputParser()

# Test Input Data
long_text = """
Machine learning is a subset of artificial intelligence that focuses on
developing systems that can learn from and make decisions based on data.
Unlike traditional programming where rules are explicitly coded, machine
learning algorithms build models based on sample data, known as training data.
These models can then make predictions or decisions without being explicitly
programmed to perform specific tasks. The field has grown tremendously with
the availability of large datasets and computational power, enabling
applications ranging from image recognition to natural language processing.
"""

print("📄 Original text:")
print(long_text)

# --- Sequential Execution Process ---

# 1. Execute Step 1: Generate summary from long text
print("\n" + "="*60)
print("STEP 1: Summarizing Content")
print("="*60)
summary_output = summarize_chain.invoke({"text": long_text})
print(summary_output)

# 2. Execute Step 2: Feed Step 1 output into the extraction chain
print("\n" + "="*60)
print("STEP 2: Extracting Key Points")
print("="*60)
# Map 'summary_output' to the '{summary}' variable in the second template
key_points = extract_chain.invoke({"summary": summary_output})
print(key_points)
```

## 2.4 More Complex Sequential Chain

```python

"""
3-step chain: Translate → Summarize → Extract sentiment
"""

# Step 1: Translate to English (if Thai)
translate_template = ChatPromptTemplate.from_messages([
    ("system", "If the text is in Thai, translate to English. If already English, return as-is."),
    ("user", "{text}")
])

# Step 2: Summarize
summarize_template = ChatPromptTemplate.from_messages([
    ("system", "Summarize in 1 sentence."),
    ("user", "{translated_text}")
])

# Step 3: Extract sentiment
sentiment_template = ChatPromptTemplate.from_messages([
    ("system", "Analyze sentiment: Positive, Negative, or Neutral. Give one word answer."),
    ("user", "{summary}")
])

# Create individual chains
translate_chain = translate_template | llm | StrOutputParser()
summarize_chain = summarize_template | llm | StrOutputParser()
sentiment_chain = sentiment_template | llm | StrOutputParser()

# Thai text example
thai_text = "วันนี้ฉันจะไปงานวันเกิดเพื่อนที่อยุธยาแต่ฉันตื่นสาย"

# Execute step by step
print("📝 Original (Thai):", thai_text)

translated = translate_chain.invoke({"text": thai_text})
print(f"\n🔄 Translated: {translated}")

summary = summarize_chain.invoke({"translated_text": translated})
print(f"\n📋 Summary: {summary}")

sentiment = sentiment_chain.invoke({"summary": summary})
print(f"\n😊 Sentiment: {sentiment}")
```

## 2.5 Parallel Chains with RunnableParallel

```python
"""
RunnableParallel: Run multiple chains at the same time!
Faster than sequential when tasks are independent
"""

# 1. Define Sub-chains (Independent tasks)
# Summary task
summary_chain = (
    ChatPromptTemplate.from_messages([
        ("system", "Summarize in one sentence."),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# Sentiment analysis task
sentiment_chain = (
    ChatPromptTemplate.from_messages([
        ("system", "What's the sentiment? Answer: Positive, Negative, or Neutral"),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# Topic extraction task
topics_chain = (
    ChatPromptTemplate.from_messages([
        ("system", "List main topics as comma-separated values."),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# 2. Combine into Parallel Execution
# Use a dictionary to map keys to their respective chains
parallel_chain = RunnableParallel({
    "summary": summary_chain,
    "sentiment": sentiment_chain,
    "topics": topics_chain
})

# Test Input
text = """
Artificial Intelligence is revolutionizing healthcare by enabling early disease
detection, personalized treatment plans, and drug discovery. Machine learning
models can analyze medical images with accuracy comparable to expert radiologists.
However, concerns about data privacy and algorithmic bias need to be addressed.
"""

# 3. Execution
# Input '{text}' is automatically distributed to all sub-chains
print(" Running 3 chains in parallel...")
results = parallel_chain.invoke({"text": text})

# 4. Access Results
# Output is a dictionary containing results from all chains
print("\n Results:")
print(f"Summary: {results['summary']}")
print(f"Sentiment: {results['sentiment']}")
print(f"Topics: {results['topics']}")
print("\n All chains executed simultaneously!")
```

## 2.6 Combining Sequential + Parallel

```python
"""
Real-world pattern: Parallel processing → Sequential aggregation
Example: Analyze from multiple angles → Combine insights
"""

# 1. Parallel Analysis Chains (Multiple Perspectives)
# Technical perspective sub-chain
technical_analysis = (
    ChatPromptTemplate.from_messages([
        ("system", "Analyze from a technical perspective. 1 sentence."),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# Business perspective sub-chain
business_analysis = (
    ChatPromptTemplate.from_messages([
        ("system", "Analyze from a business perspective. 1 sentence."),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# Ethical perspective sub-chain
ethical_analysis = (
    ChatPromptTemplate.from_messages([
        ("system", "Analyze from an ethical perspective. 1 sentence."),
        ("user", "{text}")
    ]) | llm | StrOutputParser()
)

# Group sub-chains into a parallel unit
parallel_analysis = RunnableParallel({
    "technical": technical_analysis,
    "business": business_analysis,
    "ethical": ethical_analysis
})

# 2. Synthesis Chain (Sequential Aggregation)
# This template expects the output dictionary keys from 'parallel_analysis'
synthesis_chain = (
    ChatPromptTemplate.from_messages([
        ("system", "Synthesize these three perspectives into one coherent recommendation."),
        ("user", """Technical: {technical}
Business: {business}
Ethical: {ethical}""")
    ]) | llm | StrOutputParser()
)

# 3. Complete Pipeline
# Connect Parallel processing to Sequential aggregation using the '|' operator
complete_chain = parallel_analysis | synthesis_chain

# Test
ai_text = """
A company is considering implementing facial recognition for employee attendance
tracking. The technology is accurate and would save administrative time, but
raises privacy concerns among employees.
"""

# Execute Parallel Analysis separately to show intermediate steps
print(" Step 1: Parallel analysis...")
analyses = parallel_analysis.invoke({"text": ai_text})
print(f"Technical: {analyses['technical']}")
print(f"Business: {analyses['business']}")
print(f"Ethical: {analyses['ethical']}")

# 4. Full Execution
# Data flows from Parallel Analysis -> Dictionary Output -> Synthesis Chain
print("\n Step 2: Synthesize...")
recommendation = complete_chain.invoke({"text": ai_text})
print(f"\n Final Recommendation:\n{recommendation}")
```
